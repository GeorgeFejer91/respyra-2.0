```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
```

```python
from mpi.stuff import to_viewport
```

```python
clear_output(wait=True)
plt.close("all")

amplitude = 1.0
frequency = 0.1  # Hz
tape_speed = 5.5  # unit/s
recording_time = 20  # s

t = np.linspace(0, recording_time, 400)
s = amplitude * np.sin(frequency * t)
s0 = np.min(s) - 0.2
s1 = np.max(s) + 0.2

x, y = to_viewport(t, s, s0, s1, tape_speed)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
# plt.xlim(-1, 1)
plt.ylim(-1, 1)
plt.grid(True)
plt.show()
```
